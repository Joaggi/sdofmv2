```python
import pandas as pd
```

```python
base_path = "../../../assets/test_results/"
# df_aia=pd.read_csv(base_path + "sdofmv2-aia-test-results.csv")
df_all = pd.read_csv(base_path + "sdofmv2-all-test-results-final.csv")
df_denoise = pd.read_csv(base_path + "sdofmv2-all-test-results-denoise.csv")
# df_hmi=pd.read_csv(base_path + "sdofmv2-hmi-test-results.csv")
df_all_large=pd.read_csv(base_path + "sdofmv2-large-all-test-results.csv")
df_all_small=pd.read_csv(base_path + "sdofmv2-small-all-test-results.csv")
```

```python
# Add source labels
# df_aia["dataset"] = "aia"
df_all["dataset"] = "all"
df_denoise["dataset"] = "all_denoise"
# df_hmi["dataset"] = "hmi"
df_all_large["dataset"] = "all_large"
df_all_small["dataset"] = "all_small"

# Combine
df_combined_all = pd.concat(
    [df_all, df_denoise, df_all_large, df_all_small],
    ignore_index=True
)

# Keep only relevant columns
# metrics = ["rmse_intensity", "r2_score", "pixel_correlation"]

df_long = df_combined_all.melt(
    id_vars=["dataset", "channel"],
    # value_vars=metrics,
    var_name="metric",
    value_name="value"
)

print(df_long.head())
```

```python
df_long
```

```python
aia_chan = ['131A', '1600A', '1700A', '171A', '193A', '211A', '304A', '335A', '94A']

mse_table = (
    df_long[df_long["metric"] == "mse_norm"]
    .pivot(
        index="channel",
        columns="dataset",
        values="value"
    )
)
mse_table.loc["Mean"] = mse_table.mean()
mse_table = mse_table.round(3)

rmse_table = (
    df_long[df_long["metric"] == "rmse_intensity_norm"]
    .pivot(
        index="channel",
        columns="dataset",
        values="value"
    )
)
rmse_table.loc["Mean"] = rmse_table.mean()
rmse_table = rmse_table.round(3)

mae_table = (
    df_long[df_long["metric"] == "mse_norm"]
    .pivot(
        index="channel",
        columns="dataset",
        values="value"
    )
)
mae_table.loc["Mean"] = mae_table.mean()
mae_table = mae_table.round(3)

r2_table = (
    df_long[df_long["metric"] == "r2_score_norm"]
    .pivot(
        index="channel",
        columns="dataset",
        values="value"
    )
)
r2_table.loc["Mean"] = r2_table.mean()
r2_table = r2_table.round(3)

corr_table = (
    df_long[df_long["metric"] == "pixel_correlation_norm"]
    .pivot(
        index="channel",
        columns="dataset",
        values="value"
    )
)
corr_table.loc["Mean"] = corr_table.mean()
corr_table = corr_table.round(3)
```
