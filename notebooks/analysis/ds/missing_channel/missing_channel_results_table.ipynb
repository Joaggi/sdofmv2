```python
import pandas as pd
```

```python
base_path = "../../../../assets/ds_results/missing_channel/"
columns = ["wavelength", "mse", "rmse_intensity", "mae", "r2_score"]
df_aia = pd.read_csv(base_path + "missing_channel_test_sdofmv2-aia.csv", usecols=columns)
df_all = pd.read_csv(base_path + "missing_channel_test_sdofmv2-all.csv", usecols=columns)
df_all_zero = pd.read_csv(base_path + "missing_channel_test_sdofmv2-all-zeroshot.csv", usecols=columns)
df_hmi = pd.read_csv(base_path + "missing_channel_test_sdofmv2-hmi-log.csv", usecols=columns)
```

```python
def modifying_df(df):
    df.set_index("wavelength", inplace=True)
    df.loc["mean"] = df.mean()
    df = df.rename(columns={
        "mse": "MSE",
        "rmse_intensity": "RMSE",
        "mae": "MAE",
        "r2_score": "R²"
    })
    df = df[["MSE", "RMSE", "MAE", "R²"]]
    return df
```

```python
df_aia = modifying_df(df_aia)
```

```python
# columns_order = ["mse", "rmse_intensity", "mae", "r2_score"]
df_aia
```

```python
df_all = modifying_df(df_all)
```

```python
df_all
```

```python
df_hmi = modifying_df(df_hmi)
```

```python
df_hmi
```

```python
df_all_zero = modifying_df(df_all_zero)
```

```python
df_all_zero
```

```python
df_all_combine = pd.concat([df_all, df_all_zero], axis=1)
```

```python
df_all_combine.drop(["mean", "Bx", "By", "Bz"], axis=0, inplace=True)
df_all_combine.loc["mean"] = df_all_combine.mean()
```
