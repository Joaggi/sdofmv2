```python
import zarr
from sdofmv2.utils.constants import ALL_WAVELENGTHS, ALL_COMPONENTS
```

```python
path = "/mnt/jade/AIA.zarr"
data = zarr.open_group(path, mode="r")
```

```python
total_channels = ALL_WAVELENGTHS

for year in range(2010, 2019):
    for ch in total_channels:

        # Navigate to your specific array
        array = data[f"{year}"][ch]
        print(f"Scanning array of shape {year}/{ch}: {array.shape}...")

        bad_indices = []

        # Scan through the first axis
        for i in range(array.shape[0]):
            try:
                # Attempt to load a tiny slice to trigger the chunk decompression
                _ = array[i, 0, 0] 
            except RuntimeError as e:
                if "blosc" in str(e).lower():
                    bad_indices.append(i)
                    # print(f"Corruption found at index: {i}")
            except Exception as e:
                print(f"Other error at index {i}: {e}")

        print(f"{year}/{ch}: \nTotal corrupted indices: {len(bad_indices)}")
        print(f"{year}/{ch}: List of bad indices: {bad_indices}")
```
